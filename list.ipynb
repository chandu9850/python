# @ access elements from list

 ## indexing 
Indexing means accessing elements of a list using their position number.

```python
fruits = ["apple", "banana", "mango", "orange"]
fruits
```

```python
fruits[2]
```

```python
fruits[-2]
```

```python
fruits[0]
```

```python
fruits[-4]
```

```python
abc = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]
abc
```

```python
abc[0][2]
```

```python
abc[-1][-3]
```

```python
abc[1][-2]
```

## slicing

Slicing means extracting a portion (subset) of elements from a list.

```python
a1 = [10, 20, 30, 40, 50, 60]
a1
```

```python
# 20,30,40

a1[1 : 4 : 1]
```

```python
a1[:  : -1]
```

```python
a1[:]
```

```python
# 10  │20  │30  │40  │50

a1[:5]
```

```python
a1[3:]
```

```python
a1[::2]
```

```python
a1
```
